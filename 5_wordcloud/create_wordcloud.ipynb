# Wordcloud
Dit is een wordcloud gegenereerd uit tekst gevonden op het internet met als zoekterm "halsbandparkiet" en "halsbandparkiet belgië"

```python
from collections import Counter  # Count the frequency of distinct strings
from wordcloud import WordCloud, ImageColorGenerator, STOPWORDS  # Generate wordclouds
from PIL import Image  # Load images from files
import numpy as np  # Convert images to numbers
import re
```

```python
# tekstfile inladen
with open("../resources/wordcloud/halsbandparkiet.txt", "r") as file:
    text = file.read()
    
# Woorden eruit halen en alles lower case
tokens = re.findall(r'\b\w+\b', text.lower())

# Handmatige mapping naar het enkelvoud
manual_singular_map = {
    "halsbandparkieten": "halsbandparkiet",
    "boomklevers": "boomklever",
    "soorten": "soort",
    "brusselse": "brussel",
    "groene": "groen",
    "exoten": "exoot",
}

mapped_tokens = [manual_singular_map.get(word, word) for word in tokens]

# Toevoegen van woorden die uit de weergaven moeten worden gehaald
custom_stopwords = set(STOPWORDS)
custom_stopwords.update(["het", "de", "en", "een", "van", "zijn", "op", "heeft", "voor", "te", "ze", "dat", "maar", "vooral", "geen", "hebben", "meer", "die", "je", "met", 
                         "kan", "zich", "al", "wordt", "niet", "enkele", "kunnen", "dan", "zoals", "naar", "bij", "onze", "hun", "hij", "aan",
                         "gaat", "om", "tijdens", "zoek", "alle", "wel", "dit", "komen", "s", "der", "hier", "men", "ons", "waar",
                         "e", "vaak", "toepassing", "zeer", "doen", "blijkt", "tot", "er", "soms", "uit", "onder", "zitten", "blijkt",
                         "name", "eerste", "elwin", "hoogte", "weten", "daar", "vinden", "omdat", "vormen", "kleine", "ruim", "bestaat", "zwarte", "mens", "groot", "oude", "daarnaast",
                         "natuurlijk", "staat", "p", "zo", "lijkt", "wat", "rond", "deze", "nog", "tussen", "mogelijk", "jaren", "sovon", "veel", "zien", "ontdek",
                         "komt", "moet", "door", "inmiddels", "gaan", "als", "nl", "andere", "dagen", "artikel", "werd", "mensen", "per", "jaar", "per jaar", "goed",
                         "bron", "natuurlijke", "volgen", "regel", "belangrijk", "geval", "land","blijven", "leren", "laag", "zorgen", "zwart", "relatief", "staan", "esmee",
                         "blijken", "leven", "zou", "weken", "meestal", "iedereen", "snel", "na", "vastgesteld", "echter", "werden", "bevat", "eeuw", "verschillende",
                         "worden", "grote", "geven", "toen", "zegt", "nu", "grotere", "ongeveer", "heel", "gemaakt", "zorg", "ook", "sterk", "sri", "driessens", "korte",
                         "zouden", "zij", "waren", "vanuit", "weer", "bijvoorbeeld", "maken", "sommige", "beperken",
                         "drie", "zelf", "maakt", "toch", "verder", "zonder", "regels", "blijft", "beginnen", "000", "volgens", "ten"])

filtered_tokens = [token for token in mapped_tokens if token not in custom_stopwords]

# Tel frequenties
word_counts = Counter(filtered_tokens)
print(word_counts.most_common(25))

colormap='gist_earth'

# Genereer de WordCloud
fog_machine = WordCloud(stopwords=custom_stopwords, 
                        background_color=None, mode="RGBA",
                        colormap=colormap, 
                        width=900, height=500, min_font_size=14)
fog_machine.generate(" ".join(filtered_tokens))

# Converteer naar image-object
img = fog_machine.to_image()

# Opslaan met transparante achtergrond
img.save(f"wordcloud_transparant_{colormap}.png", format="PNG")
```
